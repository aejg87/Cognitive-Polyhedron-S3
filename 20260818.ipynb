This code implements a system to discover a causal law (an equation) from a few data points, mimicking biological processes like Bayesian inference, Kolmogorov complexity (MDL principle), Bellman's principle of optimality, and simulated annealing. It aims to find a compact and accurate equation that explains the relationship between `X` and `Y_target`.

```python
import numpy as np
import random
import math

# =====================================================================
# 1. EL ENTORNO (Pocos datos / Few-Shot Learning)
# =====================================================================
X = np.array([1, 2, 3, 4, 5])
Y_target = np.array([1, 4, 7, 10, 13]) # La ley causal oculta: (X * 3) - 2

# El "ADN" disponible (Simbólico)
numeros = ['1', '2', '3']
operadores = ['+', '-', '*']

# =====================================================================
# 2. INFERENCIA BAYESIANA (El Engrama / Creencias Previas)
# =====================================================================
# El sistema no sabe qué operador es mejor al inicio (Máxima entropía)
pesos_bayes = {'+': 1.0, '-': 1.0, '*': 1.0}

def obtener_probabilidades():
    total = sum(pesos_bayes.values())
    return {k: round(v/total, 2) for k, v in pesos_bayes.items()}

def elegir_operador():
    ops = list(pesos_bayes.keys())
    probs = list(obtener_probabilidades().values())
    return random.choices(ops, weights=probs, k=1)[0]

# =====================================================================
# 3. EL JUEZ DE KOLMOGOROV (Métrica MDL / Poda Microglial)
# =====================================================================
def costo_mdl(ecuacion_str):
    try:
        x_val = X # Reemplazo para evaluación vectorial
        Y_pred = eval(ecuacion_str.replace('X', 'x_val'))
        if isinstance(Y_pred, int): Y_pred = np.full(X.shape, Y_pred)

        # 1. Costo de Error
        error_absoluto = np.sum(np.abs(Y_target - Y_pred))

        # 2. Costo de Complejidad (Poda: castiga ecuaciones largas)
        longitud = len(ecuacion_str.split())
        penalidad = longitud * 1.5

        # Energía Libre Total
        return error_absoluto + penalidad, error_absoluto
    except:
        return float('inf'), float('inf')

# =====================================================================
# 4. ORQUESTACIÓN DEL SISTEMA 3 (Bellman + Recocido Simulado)
# =====================================================================
print("INICIANDO GÉNESIS ESTRUCTURAL (SISTEMA 3)...\n")

ecuacion_actual = "X"
costo_actual, err_actual = costo_mdl(ecuacion_actual)

temperatura = 50.0  # Alta energía inicial (Expansión celular transitoria)
tasa_enfriamiento = 0.85

# Memoria de Bellman (Memoización temporal para no recalcular sub-problemas)
estados_visitados = {}

for iteracion in range(1, 101):
    if err_actual == 0:
        break # Encontró la perfección causal

    # Generar una mutación estocástica
    op_candidato = elegir_operador()
    num_candidato = random.choice(numeros)

    # Composición Dinámica de Bellman (Agrega un sub-problema al estado anterior)
    ecuacion_candidata = f"({ecuacion_actual} {op_candidato} {num_candidato})"

    if ecuacion_candidata in estados_visitados:
        continue

    costo_nuevo, err_nuevo = costo_mdl(ecuacion_candidata)
    estados_visitados[ecuacion_candidata] = costo_nuevo

    delta_costo = costo_nuevo - costo_actual
    aceptado = False

    # Evaluación y Actualización
    if delta_costo < 0:
        aceptado = True
        # Actualización Bayesiana (Ley de Hebb): Refuerza el operador exitoso
        pesos_bayes[op_candidato] += 2.0
    else:
        # Recocido Simulado: Tolerancia al error basada en Temperatura
        prob_aceptar_error = math.exp(-delta_costo / temperatura)
        if random.random() < prob_aceptar_error:
            aceptado = True
            print(f" [!] Temperatura alta ({temperatura:.1f}): Aceptando salto caótico para evitar Mínimo Local.")

    if aceptado:
        ecuacion_actual = ecuacion_candidata
        costo_actual = costo_nuevo
        err_actual = err_nuevo

    # Enfriamiento Termodinámico
    temperatura *= tasa_enfriamiento

    if iteracion % 5 == 0 or err_actual == 0:
        print(f"Iteración {iteracion:02d} | Ecuación: {ecuacion_actual: <15} | MDL: {costo_actual:.1f} | Error: {err_actual} | Pesos: {obtener_probabilidades()}")

print("\n=========================================================")
print("GÉNESIS COMPLETADA Y MEMOIZADA AL SISTEMA 1")
print("=========================================================")
print(f"Fórmula Final Comprimida : {ecuacion_actual}")
print(f"Energía Libre (Costo MDL): {costo_actual}")
print(f"Huella del Engrama (Bayes): {obtener_probabilidades()}")
```

This cell introduces an enhanced version of the causal law discovery system from the previous cell. It specifically integrates a 'Microglial Pruning' mechanism, which periodically simplifies the generated equations to prevent excessive complexity and encourage more robust, shorter solutions. This pruning is designed to reflect how biological systems might eliminate redundant or inefficient connections, leading to more generalized and stable causal models.

```python
import numpy as np
import random
import math

# =====================================================================
# 1. EL ENTORNO (Pocos datos / Few-Shot Learning)
# =====================================================================
X = np.array([1, 2, 3, 4, 5])
Y_target = np.array([1, 4, 7, 10, 13]) # La ley causal oculta: (X * 3) - 2

# El "ADN" disponible (Simbólico)
numeros = ['1', '2', '3']
operadores = ['+', '-', '*']

# =====================================================================
# 2. INFERENCIA BAYESIANA (El Engrama / Creencias Previas)
# =====================================================================
# El sistema no sabe qué operador es mejor al inicio (Máxima entropía)
pesos_bayes = {'+': 1.0, '-': 1.0, '*': 1.0}

def obtener_probabilidades():
    total = sum(pesos_bayes.values())
    return {k: round(v/total, 2) for k, v in pesos_bayes.items()}

def elegir_operador():
    ops = list(pesos_bayes.keys())
    probs = list(obtener_probabilidades().values())
    return random.choices(ops, weights=probs, k=1)[0]

# =====================================================================
# 3. EL JUEZ DE KOLMOGOROV (Métrica MDL / Poda Estructural)
# =====================================================================
def costo_mdl(ecuacion_str):
    try:
        x_val = X # Reemplazo para evaluación vectorial
        Y_pred = eval(ecuacion_str.replace('X', 'x_val'))
        if isinstance(Y_pred, int): Y_pred = np.full(X.shape, Y_pred)

        # 1. Costo de Error
        error_absoluto = np.sum(np.abs(Y_target - Y_pred))

        # 2. Costo de Complejidad (Poda: castiga fuertemente ecuaciones largas)
        longitud = len(ecuacion_str.split()) # Cuenta elementos
        penalidad = longitud * 1.5

        # Energía Libre Total
        return error_absoluto + penalidad, error_absoluto
    except:
        return float('inf'), float('inf')

# =====================================================================
# 4. ORQUESTACIÓN DEL SISTEMA 3 (Bellman + Recocido + Poda + Bayes)
# =====================================================================
print("INICIANDO GÉNESIS ESTRUCTURAL CON PODA MICROGLIAL (SISTEMA 3)...\n")

ecuacion_actual = "X"
costo_actual, err_actual = costo_mdl(ecuacion_actual)

temperatura = 50.0  # Alta energía inicial (Expansión celular transitoria)
tasa_enfriamiento = 0.85

# Memoria de Bellman (Memoización temporal para no recalcular sub-problemas)
estados_visitados = {}

for iteracion in range(1, 201): # Iteraciones suficientes para ver la biología actuar
    if err_actual == 0:
        break # Encontró la perfección causal absoluta

    # Generar una mutación estocástica
    op_candidato = elegir_operador()
    num_candidato = random.choice(numeros)

    # -----------------------------------------------------------------
    # EL NUEVO AJUSTE: Composición Dinámica con PODA SINÁPTICA
    # -----------------------------------------------------------------
    prob_poda = random.random()

    # Si la ecuación tiene más de 5 caracteres, 30% de probabilidad de podar
    if prob_poda < 0.30 and len(ecuacion_actual) > 5:
        # Poda Microglial: Destruye la cadena larga y la colapsa a su base
        ecuacion_candidata = f"(X {op_candidato} {num_candidato})"
        print(f"    [!] PODA ACTIVADA (Iteración {iteracion:02d}): Ecuación demasiado larga, destruyendo redundancia...")
    else:
        # Sinaptogénesis: Crece normalmente (Añadiendo sub-problemas de Bellman)
        ecuacion_candidata = f"({ecuacion_actual} {op_candidato} {num_candidato})"
    # -----------------------------------------------------------------

    if ecuacion_candidata in estados_visitados:
        continue

    costo_nuevo, err_nuevo = costo_mdl(ecuacion_candidata)
    estados_visitados[ecuacion_candidata] = costo_nuevo

    delta_costo = costo_nuevo - costo_actual
    aceptado = False

    # Evaluación y Actualización
    if delta_costo < 0:
        aceptado = True
        # Actualización Bayesiana (Ley de Hebb): Refuerza el operador exitoso
        pesos_bayes[op_candidato] += 2.0
    else:
        # Recocido Simulado: Tolerancia al error basada en Temperatura
        prob_aceptar_error = math.exp(-delta_costo / temperatura) if temperatura > 0.1 else 0
        if random.random() < prob_aceptar_error:
            aceptado = True

    if aceptado:
        ecuacion_actual = ecuacion_candidata
        costo_actual = costo_nuevo
        err_actual = err_nuevo

    # Enfriamiento Termodinámico
    temperatura *= tasa_enfriamiento

    if iteracion % 5 == 0 or err_actual == 0:
        print(f"Iteración {iteracion:02d} | Ecuación: {ecuacion_actual: <25} | Costo MDL: {costo_actual:.1f} | Error: {err_actual} | Pesos: {obtener_probabilidades()}")

print("\n=========================================================")
print("GÉNESIS COMPLETADA Y MEMOIZADA AL SISTEMA 1")
print("=========================================================")
print(f"Fórmula Final Comprimida : {ecuacion_actual}")
print(f"Energía Libre (Costo MDL): {costo_actual}")
print(f"Huella del Engrama (Bayes): {obtener_probabilidades()}")
```
